```python
from sklearn.metrics import average_precision_score, precision_score, recall_score, f1_score
from lightgbm import LGBMClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier
import pandas as pd
import numpy as np

print("🤖 FASE 5: PELATIHAN & AUDIT OVERFITTING 🤖")
lgb_v4 = LGBMClassifier(n_estimators=400, learning_rate=0.05, max_depth=5, num_leaves=31, scale_pos_weight=7.0, random_state=42, n_jobs=-1, verbose=-1)
xgb_v4 = XGBClassifier(n_estimators=1000, learning_rate=0.05, max_depth=5, min_child_weight=50, subsample=0.878, colsample_bytree=0.865, gamma=0.005, reg_lambda=5.0, scale_pos_weight=6.1, eval_metric='aucpr', early_stopping_rounds=40, random_state=42, n_jobs=-1)
cb_v4 = CatBoostClassifier(iterations=1000, learning_rate=0.048, depth=5, l2_leaf_reg=4.82, scale_pos_weight=7.22, eval_metric='PRAUC', random_state=42, verbose=0)

# Latih Model
print("Melatih LightGBM...")
lgb_v4.fit(X_tr_final, y_tr)
print("Melatih XGBoost...")
xgb_v4.fit(X_tr_final, y_tr, eval_set=[(X_val_final, y_val)], verbose=False)
print("Melatih CatBoost...")
cb_v4.fit(X_tr_final, y_tr, eval_set=(X_val_final, y_val), early_stopping_rounds=40, verbose=False)

# AUDIT KESEHATAN MODEL
model_dict = {'LightGBM': lgb_v4, 'XGBoost': xgb_v4, 'CatBoost': cb_v4}
audit_hasil = []

for nama, model in model_dict.items():
    pr_tr = average_precision_score(y_tr, model.predict_proba(X_tr_final)[:, 1])
    pr_va = average_precision_score(y_val, model.predict_proba(X_val_final)[:, 1])
    pr_te = average_precision_score(y_te, model.predict_proba(X_te_final)[:, 1])
    audit_hasil.append({
        'Algoritma': nama,
        'Train PR-AUC': round(pr_tr, 4),
        'Valid PR-AUC': round(pr_va, 4),
        'Test PR-AUC': round(pr_te, 4),
        'Gap (Train - Test)': round(pr_tr - pr_te, 4)
    })

print("\n📊 LAPORAN KESEHATAN MODEL (OVERFIT/UNDERFIT CHECK):")
print(pd.DataFrame(audit_hasil).to_string(index=False))

print("\n👑 FASE 6: KERJA SAMA DEWAN JURI & PENCARIAN SWEET SPOT 👑")
prob_lgb = lgb_v4.predict_proba(X_te_final)[:, 1]
prob_xgb = xgb_v4.predict_proba(X_te_final)[:, 1]
prob_cb  = cb_v4.predict_proba(X_te_final)[:, 1]

# Manual Soft Voting
y_prob_super = (prob_lgb + prob_xgb + prob_cb) / 3.0
pr_auc_super = average_precision_score(y_te, y_prob_super)
print(f"⭐ SKOR PR-AUC TEST (MANUAL ENSEMBLE): {pr_auc_super:.4f}\n")

# Scanner Titik Manis (Sweet Spot Scanner)
print("🔍 Mencari Threshold F1-Score Optimal...")
thresholds = np.arange(0.30, 0.80, 0.02)
best_f1, best_th, best_prec, best_rec = 0, 0, 0, 0

for t in thresholds:
    y_pred_t = (y_prob_super >= t).astype(int)
    f1 = f1_score(y_te, y_pred_t, zero_division=0)
    if f1 > best_f1:
        best_f1 = f1
        best_th = t
        best_prec = precision_score(y_te, y_pred_t, zero_division=0)
        best_rec = recall_score(y_te, y_pred_t, zero_division=0)

print(f"🔥 TITIK MANIS DITEMUKAN PADA THRESHOLD {best_th:.2f} 🔥")
print(f"Precision : {best_prec:.3f}")
print(f"Recall    : {best_rec:.3f}")
print(f"F1-Score  : {best_f1:.3f}")
```

## 3. Feature Engineering
Creating domain-specific features to capture advanced persistent threats, including carding/bursting behaviors (`Time_Since_Last_Txn`) and specific entity tracking (`Ultimate_UID`).

```python
# ==============================================================================
# 4. TIME-AWARE SORTING & MEGA FEATURE ENGINEERING
# ==============================================================================
# CRITICAL: Sort chronologically to prevent data leakage in time-delta features
df_train = df_train.sort_values('TransactionDT').reset_index(drop=True)

# A. Temporal & Format Features
df_train['Hari_Ke'] = df_train['TransactionDT'] // 86400
df_train['Jam_Transaksi'] = (df_train['TransactionDT'] // 3600) % 24
df_train['Hari_Dalam_Seminggu'] = df_train['Hari_Ke'] % 7
df_train['Desimal_Amt'] = df_train['TransactionAmt'] - np.floor(df_train['TransactionAmt'])

# B. Entity Tracking (Ultimate UID)
# Combining card details and address to track specific users/attackers
df_train['Ultimate_UID'] = df_train['card1'].astype(str) + '_' + df_train['card2'].astype(str) + '_' + df_train['addr1'].astype(str)

# C. Behavioral & Anomaly Features
# Calculating the time elapsed since the entity's last transaction (Catching burst attacks)
df_train['Time_Since_Last_Txn'] = df_train.groupby('Ultimate_UID')['TransactionDT'].diff().fillna(0)
df_train['UID_Freq'] = df_train.groupby('Ultimate_UID')['TransactionID'].transform('count')
df_train['Amt_per_UID_Mean'] = df_train['TransactionAmt'] / (df_train.groupby('Ultimate_UID')['TransactionAmt'].transform('mean') + 1e-5)
df_train['Total_Uang_Dihari_Itu'] = df_train.groupby(['Ultimate_UID', 'Hari_Ke'])['TransactionAmt'].cumsum()

# D. System & Missing Value Profiling
df_train['Total_NaN_Baris'] = df_train.isnull().sum(axis=1)
df_train['Email_Mismatch'] = (df_train['P_emaildomain'] != df_train['R_emaildomain']).astype(int)

# E. Financial Ratios
if 'C1' in df_train.columns and 'D1' in df_train.columns:
    df_train['Velocity_C1_D1'] = df_train['C1'] / (df_train['D1'] + 1.0)
if 'C13' in df_train.columns:
    df_train['Rasio_Amt_C13'] = df_train['TransactionAmt'] / (df_train['C13'] + 1.0)
if 'V258' in df_train.columns:
    df_train['Rasio_Amt_V258'] = df_train['TransactionAmt'] / (df_train['V258'] + 1.0)

print("Feature Engineering completed successfully.")
```

## 4. Time-Based Splitting & Preprocessing
To simulate a real-world production environment, standard cross-validation is avoided. The data is split strictly based on time (70% Train, 15% Validation, 15% Test) to ensure the model predicts the *future*, not the past.

```python
# ==============================================================================
# 5. DATA SPLITTING & ORDINAL ENCODING
# ==============================================================================
# Time-based train-validation-test split
train_idx = int(len(df_train) * 0.70)
valid_idx = int(len(df_train) * 0.85)

df_tr = df_train.iloc[:train_idx].copy()
df_val = df_train.iloc[train_idx:valid_idx].copy()
df_te = df_train.iloc[valid_idx:].copy()

# Define targets and remove metadata
kolom_buang = ['TransactionID', 'TransactionDT', 'Hari_Ke', 'isFraud']
y_tr = df_tr['isFraud']
y_val = df_val['isFraud']
y_te = df_te['isFraud']

X_tr_raw = df_tr.drop(columns=[c for c in kolom_buang if c in df_tr.columns])
X_val_raw = df_val.drop(columns=[c for c in kolom_buang if c in df_val.columns])
X_te_raw = df_te.drop(columns=[c for c in kolom_buang if c in df_te.columns])

# Algorithm-Aware Encoding: Convert categorical texts to numbers.
# Missing values (NaN) are intentionally left as-is for Gradient Boosting to handle natively.
kolom_teks = X_tr_raw.select_dtypes(include=['object']).columns.tolist()
encoder = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)

X_tr_raw[kolom_teks] = encoder.fit_transform(X_tr_raw[kolom_teks])
X_val_raw[kolom_teks] = encoder.transform(X_val_raw[kolom_teks])
X_te_raw[kolom_teks] = encoder.transform(X_te_raw[kolom_teks])

# Feature Selection Matrix
fitur_168_lama = [
    'card4', 'card6', 'P_emaildomain', 'R_emaildomain', 'M4', 'M5', 'M6', 'id_30', 'id_33', 'DeviceType', 
    'DeviceInfo', 'TransactionAmt', 'card1', 'card2', 'card3', 'card5', 'addr1', 'addr2', 'C1', 'C2', 'C4', 
    'C5', 'C6', 'C7', 'C8', 'C9', 'C10', 'C11', 'C12', 'C13', 'C14', 'D1', 'D2', 'D3', 'D4', 'D6', 'D8', 'D9', 
    'D10', 'D12', 'D14', 'D15', 'V6', 'V7', 'V13', 'V21', 'V30', 'V44', 'V45', 'V46', 'V53', 'V54', 'V56', 
    'V57', 'V58', 'V61', 'V62', 'V63', 'V67', 'V69', 'V70', 'V74', 'V75', 'V76', 'V77', 'V78', 'V80', 'V81', 
    'V83', 'V86', 'V87', 'V88', 'V93', 'V99', 'V102', 'V105', 'V123', 'V129', 'V131', 'V134', 'V135', 'V137', 
    'V139', 'V145', 'V149', 'V151', 'V152', 'V154', 'V155', 'V156', 'V159', 'V161', 'V163', 'V165', 'V166', 
    'V167', 'V169', 'V170', 'V172', 'V186', 'V188', 'V189', 'V194', 'V197', 'V199', 'V208', 'V213', 'V216', 
    'V220', 'V232', 'V242', 'V244', 'V245', 'V255', 'V257', 'V258', 'V259', 'V261', 'V267', 'V268', 'V270', 
    'V274', 'V275', 'V277', 'V279', 'V280', 'V281', 'V282', 'V283', 'V285', 'V288', 'V289', 'V291', 'V292', 
    'V293', 'V294', 'V296', 'V303', 'V304', 'V308', 'V310', 'V312', 'V313', 'V314', 'V315', 'V316', 'V317', 
    'V318', 'V321', 'V326', 'V328', 'id_01', 'id_02', 'id_06', 'id_09', 'id_17', 'id_18', 'id_19', 'id_32'
]

fitur_mega = [
    'Hari_Dalam_Seminggu', 'Ultimate_UID', 'Time_Since_Last_Txn', 'UID_Freq', 
    'Amt_per_UID_Mean', 'Velocity_C1_D1', 'Rasio_Amt_C13', 'Rasio_Amt_V258', 
    'Total_Uang_Dihari_Itu', 'Email_Mismatch', 'Total_NaN_Baris', 'Jam_Transaksi', 'Desimal_Amt'
]

# Ensure uniqueness and presence in dataset
fitur_final = list(set([f for f in (fitur_168_lama + fitur_mega) if f in X_tr_raw.columns]))

X_tr_final = X_tr_raw[fitur_final]
X_val_final = X_val_raw[fitur_final]
X_te_final = X_te_raw[fitur_final]

print(f"Data ready for modeling. Total Features: {len(fitur_final)}")
```

## 5. Model Training & Audit
Training a diverse set of Gradient Boosting algorithms. An overfit/underfit audit is conducted using PR-AUC (Precision-Recall Area Under Curve) to ensure the models maintain their predictive power on unseen data.

```python
# ==============================================================================
# 6. MODEL TRAINING & AUDIT
# ==============================================================================
# Initialize Base Models
lgb_v4 = LGBMClassifier(n_estimators=400, learning_rate=0.05, max_depth=5, num_leaves=31, scale_pos_weight=7.0, random_state=42, n_jobs=-1, verbose=-1)
xgb_v4 = XGBClassifier(n_estimators=1000, learning_rate=0.05, max_depth=5, min_child_weight=50, subsample=0.878, colsample_bytree=0.865, gamma=0.005, reg_lambda=5.0, scale_pos_weight=6.1, eval_metric='aucpr', early_stopping_rounds=40, random_state=42, n_jobs=-1)
cb_v4 = CatBoostClassifier(iterations=1000, learning_rate=0.048, depth=5, l2_leaf_reg=4.82, scale_pos_weight=7.22, eval_metric='PRAUC', random_state=42, verbose=0)

# Train Models
print("Training LightGBM...")
lgb_v4.fit(X_tr_final, y_tr)
print("Training XGBoost...")
xgb_v4.fit(X_tr_final, y_tr, eval_set=[(X_val_final, y_val)], verbose=False)
print("Training CatBoost...")
cb_v4.fit(X_tr_final, y_tr, eval_set=(X_val_final, y_val), early_stopping_rounds=40, verbose=False)

# Overfit/Underfit Audit
model_dict = {'LightGBM': lgb_v4, 'XGBoost': xgb_v4, 'CatBoost': cb_v4}
audit_hasil = []

for nama, model in model_dict.items():
    pr_tr = average_precision_score(y_tr, model.predict_proba(X_tr_final)[:, 1])
    pr_va = average_precision_score(y_val, model.predict_proba(X_val_final)[:, 1])
    pr_te = average_precision_score(y_te, model.predict_proba(X_te_final)[:, 1])
    audit_hasil.append({
        'Algorithm': nama,
        'Train PR-AUC': round(pr_tr, 4),
        'Valid PR-AUC': round(pr_va, 4),
        'Test PR-AUC': round(pr_te, 4),
        'Gap (Train-Test)': round(pr_tr - pr_te, 4)
    })

print("\n--- MODEL HEALTH REPORT ---")
print(pd.DataFrame(audit_hasil).to_string(index=False))
```

## 6. Super Ensemble & Sweet Spot Scanning
Merging predictions via Manual Soft Voting (1:1:1) to boost robustness. A threshold scanner is deployed to find the optimal decision boundary that balances Precision (reducing false alarms) and Recall (catching the fraud).

```python
# ==============================================================================
# 7. ENSEMBLE & THRESHOLD OPTIMIZATION
# ==============================================================================
# Extract Probabilities
prob_lgb = lgb_v4.predict_proba(X_te_final)[:, 1]
prob_xgb = xgb_v4.predict_proba(X_te_final)[:, 1]
prob_cb  = cb_v4.predict_proba(X_te_final)[:, 1]

# Manual Soft Voting Ensemble
y_prob_super = (prob_lgb + prob_xgb + prob_cb) / 3.0
pr_auc_super = average_precision_score(y_te, y_prob_super)
print(f"ENSEMBLE PR-AUC SCORE: {pr_auc_super:.4f}\n")

# Sweet Spot Scanner (Threshold Tuning)
thresholds = np.arange(0.30, 0.80, 0.02)
best_f1, best_th, best_prec, best_rec = 0, 0, 0, 0

for t in thresholds:
    y_pred_t = (y_prob_super >= t).astype(int)
    f1 = f1_score(y_te, y_pred_t, zero_division=0)
    if f1 > best_f1:
        best_f1 = f1
        best_th = t
        best_prec = precision_score(y_te, y_pred_t, zero_division=0)
        best_rec = recall_score(y_te, y_pred_t, zero_division=0)

print("--- BUSINESS IMPACT METRICS ---")
print(f"Optimal Threshold : {best_th:.2f}")
print(f"Precision         : {best_prec:.3f}")
print(f"Recall            : {best_rec:.3f}")
print(f"F1-Score          : {best_f1:.3f}")
```

## 6.5. Confusion Matrix (Business Translation)
Translating the optimal threshold (0.66) into absolute business numbers. This visualizes exactly how many legitimate transactions were flagged as false alarms, and how many fraudulent transactions were successfully blocked.

```python
# ==============================================================================
# 7.5. CONFUSION MATRIX VISUALIZATION
# ==============================================================================
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Generate predictions using the optimal threshold discovered earlier
y_pred_best = (y_prob_super >= best_th).astype(int)

# Calculate confusion matrix
cm = confusion_matrix(y_te, y_pred_best)

# Plotting
plt.figure(figsize=(7, 5))
ax = sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                 xticklabels=['Legitimate (0)', 'Fraud (1)'],
                 yticklabels=['Legitimate (0)', 'Fraud (1)'],
                 annot_kws={"size": 14})

plt.title(f"Confusion Matrix (Threshold: {best_th:.2f})\nPrioritizing Precision & CX", fontsize=14, pad=15)
plt.ylabel('Actual Transaction', fontsize=12)
plt.xlabel('AI Prediction', fontsize=12)

# Menambahkan penjelasan detail di bawah grafik
plt.figtext(0.5, -0.05, 
            f"True Negatives (Aman): {cm[0,0]} | False Positives (Salah Blokir/Alarm Palsu): {cm[0,1]}\n"
            f"False Negatives (Maling Lolos): {cm[1,0]} | True Positives (Maling Tertangkap): {cm[1,1]}", 
            ha="center", fontsize=11, bbox={"facecolor":"orange", "alpha":0.2, "pad":5})

plt.show()
```

## 7. Explainable AI (SHAP Analysis)
In the highly regulated financial sector, "black-box" AI models are unacceptable. Stakeholders and auditors need to know *why* a transaction was blocked. 

Using **SHAP (SHapley Additive exPlanations)**, we extract the decision-making logic of our LightGBM model to understand which engineered features have the strongest impact on detecting fraud.

```python
# ==============================================================================
# 8. SHAP ANALYSIS (MODEL EXPLAINABILITY)
# ==============================================================================
import shap
import matplotlib.pyplot as plt

print("Memindai otak AI dengan SHAP (Mohon tunggu beberapa detik)...")

# Mengambil sampel 2000 baris dari Test Set agar komputasi ringan dan cepat
X_shap_sample = X_te_final[:2000]

# Inisialisasi explainer khusus untuk LightGBM
explainer = shap.TreeExplainer(lgb_v4)
shap_values = explainer.shap_values(X_shap_sample)

# Ekstraksi nilai probabilitas untuk kelas 1 (Fraud)
if isinstance(shap_values, list):
    shap_values_fraud = shap_values[1]
else:
    shap_values_fraud = shap_values

# Visualisasi SHAP Summary Plot
shap.initjs()
plt.figure(figsize=(10, 8))
plt.title("SHAP Summary Plot: Top Fraud Drivers (Mega FE)", fontsize=14)

shap.summary_plot(
    shap_values_fraud, 
    X_shap_sample, 
    feature_names=fitur_final,
    show=False
)
plt.tight_layout()
plt.show()
```
